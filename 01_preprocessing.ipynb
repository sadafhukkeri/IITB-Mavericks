```python
folders = {
    "EEG": "../data/EEG",
    "GSR": "../data/GSR",
    "PSY": "../data/PSY"
}

num_files = 38

for folder_name, folder_path in folders.items():
    if not os.path.exists(folder_path):
        print(f"[WARN] Folder not found: {folder_path}")
        continue

    print(f"\n=== Processing folder: {folder_name} ===")

    for i in range(1, num_files + 1):
        file_name = f"{i}_{folder_name}.csv"
        file_path = os.path.join(folder_path, file_name)
        
        if not os.path.exists(file_path):
            print(f"[WARN] File not found: {file_name}")
            continue
        
        df = pd.read_csv(file_path)
        
        if folder_name == "EEG":
            # Fill all EEG columns with mean
            df.fillna(df.mean(), inplace=True)
        
        elif folder_name == "GSR":
            # Fill GSR columns with linear interpolation
            df.interpolate(method='linear', inplace=True)
            df.fillna(df.mean(), inplace=True)  # in case first/last rows are NaN
        
        elif folder_name == "PSY":
            # Fill timestamps with forward fill, Difficulty with median
            for col in df.columns:
                if df[col].dtype == 'object':
                    df[col].fillna(method='ffill', inplace=True)
                else:
                    df[col].fillna(df[col].median(), inplace=True)
        
        # Save the cleaned file
        df.to_csv(file_path, index=False)
        print(f"{file_name} cleaned and saved.")
```

```python
eeg_path = "../data/EEG"
gsr_path = "../data/GSR"
psy_path = "../data/PSY"
num_files = 38

# Initialize master dataset
all_trials = []

for i in range(1, num_files + 1):
    print(f"Processing participant {i}...")

    # Load data
    eeg_file = os.path.join(eeg_path, f"{i}_EEG.csv")
    gsr_file = os.path.join(gsr_path, f"{i}_GSR.csv")
    psy_file = os.path.join(psy_path, f"{i}_PSY.csv")
    
    if not (os.path.exists(eeg_file) and os.path.exists(gsr_file) and os.path.exists(psy_file)):
        print(f"[WARN] Missing files for participant {i}, skipping.")
        continue
    
    eeg_df = pd.read_csv(eeg_file)
    gsr_df = pd.read_csv(gsr_file)
    psy_df = pd.read_csv(psy_file)
    
    # Drop empty columns if any
    eeg_df.dropna(how='all', inplace=True)
    gsr_df.dropna(how='all', inplace=True)
    
    # Total duration of the session
    session_start = psy_df['routineStart'].min()
    session_end   = psy_df['routineEnd'].max()
    total_duration = session_end - session_start
    
    n_eeg = len(eeg_df)
    n_gsr = len(gsr_df)
    
    # Loop over each trial in PSY
    for idx, trial in psy_df.iterrows():
        trial_start = trial['routineStart']
        trial_end   = trial['routineEnd']
        label = trial['Difficulty']  # Map to binary or 3-class if needed

        # Map PSY trial to EEG rows proportionally
        start_idx_eeg = int((trial_start - session_start) / total_duration * n_eeg)
        end_idx_eeg   = int((trial_end - session_start) / total_duration * n_eeg)
        eeg_trial = eeg_df.iloc[start_idx_eeg:end_idx_eeg]
        
        # Map PSY trial to GSR rows proportionally
        start_idx_gsr = int((trial_start - session_start) / total_duration * n_gsr)
        end_idx_gsr   = int((trial_end - session_start) / total_duration * n_gsr)
        gsr_trial = gsr_df.iloc[start_idx_gsr:end_idx_gsr]
        
        # Skip trial if no data
        if eeg_trial.empty or gsr_trial.empty:
            continue
        
        features = {}
        
        # Aggregate EEG features
        for col in eeg_trial.columns:
            features[f"{col}_mean"] = eeg_trial[col].mean()
            features[f"{col}_std"] = eeg_trial[col].std()
            features[f"{col}_min"] = eeg_trial[col].min()
            features[f"{col}_max"] = eeg_trial[col].max()
            features[f"{col}_median"] = eeg_trial[col].median()
        
        # Aggregate GSR features
        for col in gsr_trial.columns:
            features[f"{col}_mean"] = gsr_trial[col].mean()
            features[f"{col}_std"] = gsr_trial[col].std()
            features[f"{col}_min"] = gsr_trial[col].min()
            features[f"{col}_max"] = gsr_trial[col].max()
            features[f"{col}_median"] = gsr_trial[col].median()
        
        # Add participant, trial, and label
        features['participant'] = i
        features['trial'] = idx + 1
        features['label'] = label
        
        all_trials.append(features)

# Create master DataFrame
master_df = pd.DataFrame(all_trials)

# Fill any remaining NaNs with 0
master_df.fillna(0, inplace=True)

output_file = os.path.join("data", "per_trial_dataset.csv")
master_df.to_csv(output_file, index=False)
print(f"Per-trial dataset created: {output_file}")
```

```python
eeg_folder = "../data/EEG"
gsr_folder = "../data/GSR"
output_folder = "../data/merged_data"
os.makedirs(output_folder, exist_ok=True)

num_students = 38

for i in range(1, num_students + 1):
    print(f"Merging EEG + GSR for student {i}...")

    eeg_file = os.path.join(eeg_folder, f"{i}_EEG.csv")
    gsr_file = os.path.join(gsr_folder, f"{i}_GSR.csv")
    
    if not (os.path.exists(eeg_file) and os.path.exists(gsr_file)):
        print(f"[WARN] Missing EEG or GSR file for student {i}, skipping.")
        continue
    
    # Load files
    eeg_df = pd.read_csv(eeg_file)
    gsr_df = pd.read_csv(gsr_file)
    
    # Merge based on UnixTime (nearest match)
    if "UnixTime" in eeg_df.columns and "UnixTime" in gsr_df.columns:
        merged_df = pd.merge_asof(
            eeg_df.sort_values("UnixTime"),
            gsr_df.sort_values("UnixTime"),
            on="UnixTime",
            direction="nearest"
        )
    else:
        merged_df = pd.concat([eeg_df, gsr_df], axis=1)
    
    # Save merged CSV
    output_file = os.path.join(output_folder, f"student_{i}.csv")
    merged_df.to_csv(output_file, index=False)

print("All students merged and saved in 'merged_data' folder.")
```

```python
merged_folder = "../data/merged_data"  # merged EEG+GSR files
psy_folder = "../data/PSY"             # PSY metadata (1_PSY.csv ... 38_PSY.csv)
output_file = "../data/per_trial_dataset.csv"

num_students = 38
all_trials = []

for i in range(1, num_students + 1):
    print(f"Processing participant {i}...")

    # Load merged EEG+GSR
    merged_file = os.path.join(merged_folder, f"student_{i}.csv")
    psy_file = os.path.join(psy_folder, f"{i}_PSY.csv")  # <- corrected path
    
    if not (os.path.exists(merged_file) and os.path.exists(psy_file)):
        print(f"[WARN] Missing files for participant {i}, skipping.")
        continue
    
    merged_df = pd.read_csv(merged_file)
    psy_df = pd.read_csv(psy_file)

    # Ensure Difficulty numeric
    psy_df['Difficulty'] = pd.to_numeric(psy_df['Difficulty'], errors='coerce')
    psy_df.dropna(subset=['Difficulty'], inplace=True)
    
    if len(psy_df) == 0:
        continue

    # Iterate over trials
    for idx, trial in psy_df.iterrows():
        start, end = trial["routineStart"], trial["routineEnd"]
        label = trial["Difficulty"]
        key = trial.get("Key", idx + 1)

        # Slice merged data for trial window
        segment = merged_df[(merged_df["UnixTime"] >= start) & (merged_df["UnixTime"] <= end)]
        if segment.empty:
            continue

        features = {}

        # Numeric columns (EEG + GSR)
        numeric_cols = segment.select_dtypes(include=np.number).columns
        numeric_cols = [c for c in numeric_cols if c not in ["UnixTime", "Timestamp"]]

        # Aggregate features: mean, std, min, max, median
        for col in numeric_cols:
            features[f"{col}_mean"] = segment[col].mean()
            features[f"{col}_std"] = segment[col].std()
            features[f"{col}_min"] = segment[col].min()
            features[f"{col}_max"] = segment[col].max()
            features[f"{col}_median"] = segment[col].median()

        # Add participant, trial, label
        features['participant'] = i
        features['trial'] = key
        features['label'] = label

        all_trials.append(features)

# Create master DataFrame
master_df = pd.DataFrame(all_trials)
master_df.fillna(0, inplace=True)

# Save CSV
master_df.to_csv(output_file, index=False)
print(f"Per-trial dataset created: {output_file}")
```

```python
import pandas as pd
import numpy as np
import os
import logging
from pathlib import Path
from typing import Dict, List, Optional, Tuple

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('feature_extraction.log'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

class EEGGSRFeatureExtractor:
    """Extract per-trial features from EEG+GSR and PSY data files with timestamp synchronization."""
    
    def __init__(self, merged_data_dir: str = "../data/merged_data", psy_data_dir: str = "../data/PSY", 
                 output_dir: str = "../data", target_sampling_rate: float = 250.0):
        self.merged_data_dir = Path(merged_data_dir)
        self.psy_data_dir = Path(psy_data_dir)
        self.output_dir = Path(output_dir)
        self.output_dir.mkdir(exist_ok=True)
        self.target_sampling_rate = target_sampling_rate  # Target sampling rate for alignment

        # EEG frequency bands (Hz)
        self.eeg_bands = {
            'Delta': (0.5, 4),
            'Theta': (4, 8),
            'Alpha': (8, 13),
            'Beta': (13, 30),
            'Gamma': (30, 100)
        }

        # Statistical functions to compute
        self.stat_functions = {
            'mean': np.mean,
            'std': np.std,
            'min': np.min,
            'max': np.max,
            'median': np.median
        }

    def identify_eeg_gsr_columns(self, df: pd.DataFrame) -> Tuple[List[str], List[str]]:
        """Identify EEG and GSR columns from the dataframe."""
        eeg_columns = []
        gsr_columns = []
        for col in df.columns:
            col_lower = col.lower()
            if any(eeg_indicator in col_lower for eeg_indicator in ['eeg', 'electrode', 'channel']):
                eeg_columns.append(col)
            elif any(gsr_indicator in col_lower for gsr_indicator in ['gsr', 'eda', 'conductance']):
                gsr_columns.append(col)
            # Additional heuristics - assume numeric columns that aren't time-related are EEG
            elif col not in ['timestamp', 'time', 'Timestamp', 'Time', 'sample', 'index'] and pd.api.types.is_numeric_dtype(df[col]):
                if 'gsr' in col_lower or 'eda' in col_lower or 'conductance' in col_lower:
                    gsr_columns.append(col)
                else:
                    eeg_columns.append(col)
        logger.info(f"Identified {len(eeg_columns)} EEG columns and {len(gsr_columns)} GSR columns")
        return eeg_columns, gsr_columns

    def synchronize_timestamps(self, merged_df: pd.DataFrame) -> pd.DataFrame:
        """Synchronize and standardize timestamps in the merged dataframe."""
        
        # Find timestamp column
        timestamp_col = None
        for col in ['timestamp', 'time', 'Timestamp', 'Time', 'sample_time']:
            if col in merged_df.columns:
                timestamp_col = col
                break
        
        if timestamp_col is None:
            logger.warning("No timestamp column found, creating synthetic timestamps")
            merged_df = merged_df.copy()
            merged_df['timestamp'] = np.arange(len(merged_df)) / self.target_sampling_rate
            timestamp_col = 'timestamp'
            return merged_df
        
        logger.info(f"Synchronizing timestamps using column: {timestamp_col}")
        
        # Convert timestamp column to numeric if needed
        if not pd.api.types.is_numeric_dtype(merged_df[timestamp_col]):
            logger.info(f"Converting {timestamp_col} from {merged_df[timestamp_col].dtype} to numeric")
            
            # Check if timestamps are datetime strings
            sample_timestamp = str(merged_df[timestamp_col].iloc[0])
            if '-' in sample_timestamp and ':' in sample_timestamp:
                logger.info("Converting datetime strings to Unix timestamps")
                try:
                    merged_df[timestamp_col] = pd.to_datetime(merged_df[timestamp_col]).astype('int64') / 1e9
                    logger.info("Successfully converted datetime strings to Unix timestamps")
                except Exception as e:
                    logger.error(f"Error converting datetime strings: {e}")
                    return merged_df
            else:
                merged_df[timestamp_col] = pd.to_numeric(merged_df[timestamp_col], errors='coerce')
        
        # Check for and handle duplicate or out-of-order timestamps
        if merged_df[timestamp_col].duplicated().any():
            logger.warning("Found duplicate timestamps - removing duplicates")
            merged_df = merged_df.drop_duplicates(subset=[timestamp_col], keep='first')
        
        if not merged_df[timestamp_col].is_monotonic_increasing:
            logger.warning("Timestamps are not in chronological order - sorting by timestamp")
            merged_df = merged_df.sort_values(timestamp_col).reset_index(drop=True)
        
        # Remove any NaN timestamps
        if merged_df[timestamp_col].isna().any():
            logger.warning("Removing rows with NaN timestamps")
            merged_df = merged_df.dropna(subset=[timestamp_col])
        
        logger.info(f"Timestamp synchronization complete. Data shape: {merged_df.shape}")
        return merged_df
    
    def estimate_sampling_rate(self, timestamps: pd.Series) -> float:
        """Estimate the sampling rate from timestamp differences."""
        if len(timestamps) < 2:
            return self.target_sampling_rate
        
        # Calculate time differences between consecutive samples
        time_diffs = timestamps.diff().dropna()
        
        # Remove outliers (keep values within 3 standard deviations)
        mean_diff = time_diffs.mean()
        std_diff = time_diffs.std()
        valid_diffs = time_diffs[abs(time_diffs - mean_diff) <= 3 * std_diff]
        
        if len(valid_diffs) == 0:
            return self.target_sampling_rate
        
        # Calculate sampling rate as 1 / mean_time_difference
        mean_time_diff = valid_diffs.mean()
        estimated_rate = 1.0 / mean_time_diff if mean_time_diff > 0 else self.target_sampling_rate
        
        logger.info(f"Estimated sampling rate: {estimated_rate:.2f} Hz")
        return estimated_rate
    
    def resample_modality_data(self, data: pd.DataFrame, timestamps: pd.Series, 
                              target_rate: float, method: str = 'linear') -> pd.DataFrame:
        """Resample data to target sampling rate for modality alignment."""
        
        if len(data) != len(timestamps):
            logger.error("Data and timestamps length mismatch")
            return data
        
        if len(data) < 2:
            logger.warning("Insufficient data for resampling")
            return data
        
        # Estimate current sampling rate
        current_rate = self.estimate_sampling_rate(timestamps)
        
        if abs(current_rate - target_rate) < 1.0:  # If rates are close, don't resample
            logger.info(f"Sampling rates are close ({current_rate:.2f} vs {target_rate:.2f}), skipping resample")
            return data
        
        logger.info(f"Resampling from {current_rate:.2f} Hz to {target_rate:.2f} Hz using {method} interpolation")
        
        try:
            # Create new timestamp array at target sampling rate
            start_time = timestamps.iloc[0]
            end_time = timestamps.iloc[-1]
            duration = end_time - start_time
            n_new_samples = int(duration * target_rate)
            
            new_timestamps = np.linspace(start_time, end_time, n_new_samples)
            
            # Resample each column
            resampled_data = pd.DataFrame(index=range(n_new_samples))
            resampled_data['timestamp'] = new_timestamps
            
            for col in data.columns:
                if col in ['timestamp', 'time', 'Timestamp', 'Time']:
                    continue  # Skip timestamp columns
                    
                if pd.api.types.is_numeric_dtype(data[col]):
                    # Interpolate numeric data
                    resampled_data[col] = np.interp(new_timestamps, timestamps, data[col])
                else:
                    # For non-numeric data, use forward fill
                    logger.warning(f"Forward filling non-numeric column: {col}")
                    resampled_data[col] = data[col].iloc[0]  # Use first value
            
            logger.info(f"Resampling complete: {len(data)} -> {len(resampled_data)} samples")
            return resampled_data
            
        except Exception as e:
            logger.error(f"Error during resampling: {e}")
            return data
    
    def align_modalities(self, trial_data: pd.DataFrame, eeg_columns: List[str], 
                        gsr_columns: List[str]) -> pd.DataFrame:
        """Align EEG and GSR modalities by resampling to common sampling rate."""
        
        timestamp_col = None
        for col in ['timestamp', 'time', 'Timestamp', 'Time']:
            if col in trial_data.columns:
                timestamp_col = col
                break
        
        if timestamp_col is None:
            logger.warning("No timestamp column found for modality alignment")
            return trial_data
        
        logger.info("Aligning EEG and GSR modalities")
        
        # Separate EEG and GSR data
        eeg_data = trial_data[[timestamp_col] + eeg_columns].copy()
        gsr_data = trial_data[[timestamp_col] + gsr_columns].copy()
        other_data = trial_data.drop(columns=eeg_columns + gsr_columns)
        
        # Estimate sampling rates for each modality
        eeg_rate = self.estimate_sampling_rate(eeg_data[timestamp_col]) if len(eeg_columns) > 0 else self.target_sampling_rate
        gsr_rate = self.estimate_sampling_rate(gsr_data[timestamp_col]) if len(gsr_columns) > 0 else self.target_sampling_rate
        
        logger.info(f"EEG sampling rate: {eeg_rate:.2f} Hz, GSR sampling rate: {gsr_rate:.2f} Hz")
        
        # Choose target sampling rate (use the higher of the two, or target rate)
        target_rate = max(eeg_rate, gsr_rate, self.target_sampling_rate)
        target_rate = min(target_rate, 1000.0)  # Cap at 1000 Hz for practical reasons
        
        logger.info(f"Target alignment sampling rate: {target_rate:.2f} Hz")
        
        # Resample EEG data if needed
        if len(eeg_columns) > 0 and abs(eeg_rate - target_rate) > 1.0:
            eeg_data = self.resample_modality_data(eeg_data, eeg_data[timestamp_col], target_rate)
        
        # Resample GSR data if needed
        if len(gsr_columns) > 0 and abs(gsr_rate - target_rate) > 1.0:
            gsr_data = self.resample_modality_data(gsr_data, gsr_data[timestamp_col], target_rate)
        
        # Merge the aligned data back together
        try:
            if len(eeg_columns) > 0 and len(gsr_columns) > 0:
                # Merge EEG and GSR data on timestamp
                aligned_data = pd.merge_asof(
                    eeg_data.sort_values(timestamp_col), 
                    gsr_data.sort_values(timestamp_col), 
                    on=timestamp_col, 
                    direction='nearest',
                    tolerance=1.0/target_rate  # Allow small timestamp differences
                )
            elif len(eeg_columns) > 0:
                aligned_data = eeg_data
            elif len(gsr_columns) > 0:
                aligned_data = gsr_data
            else:
                aligned_data = trial_data
            
            logger.info(f"Modality alignment complete: {len(trial_data)} -> {len(aligned_data)} samples")
            return aligned_data
            
        except Exception as e:
            logger.error(f"Error during modality alignment: {e}")
            return trial_data
        """Identify EEG and GSR columns from the dataframe."""
        eeg_columns = []
        gsr_columns = []
        
        for col in df.columns:
            col_lower = col.lower()
            if any(eeg_indicator in col_lower for eeg_indicator in ['eeg', 'electrode', 'channel']):
                eeg_columns.append(col)
            elif any(gsr_indicator in col_lower for gsr_indicator in ['gsr', 'eda', 'conductance']):
                gsr_columns.append(col)
            # Additional heuristics - assume numeric columns that aren't time-related are EEG
            elif col not in ['timestamp', 'time', 'Timestamp', 'Time', 'sample', 'index'] and pd.api.types.is_numeric_dtype(df[col]):
                # If we don't have explicit EEG/GSR indicators, assume most numeric columns are EEG
                if 'gsr' in col_lower or 'eda' in col_lower or 'conductance' in col_lower:
                    gsr_columns.append(col)
                else:
                    eeg_columns.append(col)
        
        logger.info(f"Identified {len(eeg_columns)} EEG columns and {len(gsr_columns)} GSR columns")
        return eeg_columns, gsr_columns
    
    def compute_basic_features(self, data: pd.Series, prefix: str) -> Dict[str, float]:
        """Compute basic statistical features for a data series."""
        features = {}
        
        if len(data) == 0 or data.isna().all():
            # Handle empty or all-NaN data
            for stat_name in self.stat_functions.keys():
                features[f"{prefix}_{stat_name}"] = np.nan
            return features
        
        # Remove NaN values for computation
        clean_data = data.dropna()
        
        if len(clean_data) == 0:
            for stat_name in self.stat_functions.keys():
                features[f"{prefix}_{stat_name}"] = np.nan
            return features
        
        for stat_name, stat_func in self.stat_functions.items():
            try:
                features[f"{prefix}_{stat_name}"] = stat_func(clean_data)
            except Exception as e:
                logger.warning(f"Error computing {stat_name} for {prefix}: {e}")
                features[f"{prefix}_{stat_name}"] = np.nan
        
        return features
    
    def find_time_alignment(self, merged_df: pd.DataFrame, psy_df: pd.DataFrame, 
                           timestamp_col: str) -> float:
        """Find the time offset between PSY and merged data timestamps."""
        
        # First, ensure merged data timestamps are converted to numeric
        merged_df_copy = merged_df.copy()
        
        # Convert timestamp column if needed
        if not pd.api.types.is_numeric_dtype(merged_df_copy[timestamp_col]):
            logger.info("Converting merged data timestamps for alignment calculation")
            sample_timestamp = str(merged_df_copy[timestamp_col].iloc[0])
            if '-' in sample_timestamp and ':' in sample_timestamp:
                try:
                    merged_df_copy[timestamp_col] = pd.to_datetime(merged_df_copy[timestamp_col]).astype('int64') / 1e9
                except Exception as e:
                    logger.error(f"Error converting timestamps for alignment: {e}")
                    return 0.0
            else:
                merged_df_copy[timestamp_col] = pd.to_numeric(merged_df_copy[timestamp_col], errors='coerce')
        
        # Get PSY timestamp range
        psy_start_cols = ['routineStart', 'routine_start', 'start_time', 'StartTime']
        psy_end_cols = ['routineEnd', 'routine_end', 'end_time', 'EndTime']
        
        psy_starts = []
        psy_ends = []
        
        for _, trial in psy_df.iterrows():
            for col in psy_start_cols:
                if col in trial.index and pd.notna(trial[col]):
                    try:
                        psy_starts.append(float(trial[col]))
                        break
                    except:
                        continue
                        
            for col in psy_end_cols:
                if col in trial.index and pd.notna(trial[col]):
                    try:
                        psy_ends.append(float(trial[col]))
                        break
                    except:
                        continue
        
        if not psy_starts or not psy_ends:
            logger.error("No valid PSY timestamps found")
            return 0.0
            
        psy_min = min(psy_starts)
        psy_max = max(psy_ends)
        
        # Get merged data timestamp range (now guaranteed to be numeric)
        data_min = merged_df_copy[timestamp_col].min()
        data_max = merged_df_copy[timestamp_col].max()
        
        logger.info(f"PSY timestamp range: {float(psy_min):.2f} - {float(psy_max):.2f}")
        logger.info(f"Merged data timestamp range: {float(data_min):.2f} - {float(data_max):.2f}")
        
        # Calculate potential offsets
        offset_start = float(data_min) - float(psy_min)
        offset_end = float(data_max) - float(psy_max)
        
        logger.info(f"Potential time offsets: start={offset_start:.2f}s, end={offset_end:.2f}s")
        
        # Use the average offset
        offset = (offset_start + offset_end) / 2
        
        # Check if offset makes sense (should be reasonable)
        if abs(offset) > 86400:  # More than 24 hours difference
            logger.warning(f"Large time offset detected: {float(offset):.2f}s ({float(offset)/3600:.2f} hours)")
        
        logger.info(f"Using time offset: {float(offset):.2f}s ({float(offset)/3600:.2f} hours)")
        return float(offset)
        """Compute EEG band-specific features using simple frequency domain analysis."""
        features = {}
        
        if eeg_data.empty:
            for band_name in self.eeg_bands.keys():
                for stat_name in self.stat_functions.keys():
                    features[f"EEG_{band_name}_{stat_name}"] = np.nan
            return features
        
        try:
            from scipy import signal
            from scipy.fft import fft, fftfreq
            
            for col in eeg_data.columns:
                data = eeg_data[col].dropna()
                
                if len(data) < 10:  # Need minimum samples for FFT
                    continue
                
                # Compute FFT
                n = len(data)
                fft_vals = fft(data)
                freqs = fftfreq(n, 1/sampling_rate)
                
                # Compute power spectral density
                psd = np.abs(fft_vals) ** 2
                
                # Extract features for each frequency band
                for band_name, (low_freq, high_freq) in self.eeg_bands.items():
                    # Find frequency indices for this band
                    band_mask = (freqs >= low_freq) & (freqs <= high_freq)
                    
                    if np.sum(band_mask) == 0:
                        continue
                    
                    band_power = psd[band_mask]
                    
                    # Compute statistics for this band
                    for stat_name, stat_func in self.stat_functions.items():
                        feature_name = f"EEG_{col}_{band_name}_{stat_name}"
                        try:
                            features[feature_name] = stat_func(band_power)
                        except:
                            features[feature_name] = np.nan
        
        except ImportError:
            logger.warning("scipy not available - skipping EEG band analysis")
        except Exception as e:
            logger.warning(f"Error in EEG band analysis: {e}")
        
        return features
    
    def load_files(self, participant_id: int) -> Tuple[Optional[pd.DataFrame], Optional[pd.DataFrame]]:
        """Load merged EEG+GSR and PSY files for a participant."""
        # Load merged EEG+GSR file
        merged_file = self.merged_data_dir / f"student_{participant_id}.csv"
        merged_df = None
        
        if merged_file.exists():
            try:
                merged_df = pd.read_csv(merged_file, low_memory=False)
                logger.info(f"Loaded merged data for participant {participant_id}: {merged_df.shape}")
            except Exception as e:
                logger.error(f"Error loading {merged_file}: {e}")
        else:
            logger.warning(f"Merged file not found: {merged_file}")
        
        # Load PSY file
        psy_file = self.psy_data_dir / f"{participant_id}_PSY.csv"
        psy_df = None
        
        if psy_file.exists():
            try:
                psy_df = pd.read_csv(psy_file)
                logger.info(f"Loaded PSY data for participant {participant_id}: {psy_df.shape}")
                
                # Debug: Show the first few rows and column names
                logger.debug(f"PSY columns: {list(psy_df.columns)}")
                logger.debug(f"First PSY row: {psy_df.iloc[0].to_dict() if len(psy_df) > 0 else 'Empty'}")
                
            except Exception as e:
                logger.error(f"Error loading {psy_file}: {e}")
        else:
            logger.warning(f"PSY file not found: {psy_file}")
        
        return merged_df, psy_df
    
    def slice_trial_data(self, merged_df: pd.DataFrame, routine_start: float, 
                        routine_end: float, time_offset: float = 0.0, timestamp_col: str = 'timestamp') -> pd.DataFrame:
        """Slice the merged data for a specific trial based on routine start/end times with proper synchronization."""
        
        logger.info(f"Slicing trial data for time range: {routine_start:.2f} - {routine_end:.2f}")
        
        # Synchronize timestamps first
        synchronized_df = self.synchronize_timestamps(merged_df)
        
        # Find the actual timestamp column after synchronization
        actual_timestamp_col = timestamp_col
        if timestamp_col not in synchronized_df.columns:
            for col in ['timestamp', 'time', 'Timestamp', 'Time', 'sample_time']:
                if col in synchronized_df.columns:
                    actual_timestamp_col = col
                    break
        
        try:
            # Apply time offset to align PSY and merged data timestamps
            adjusted_start = routine_start + time_offset
            adjusted_end = routine_end + time_offset
            
            # Get timestamp range information
            data_min = synchronized_df[actual_timestamp_col].min()
            data_max = synchronized_df[actual_timestamp_col].max()
            
            logger.info(f"Data timestamp range: {float(data_min):.2f} - {float(data_max):.2f}")
            logger.info(f"Adjusted search range: {float(adjusted_start):.2f} - {float(adjusted_end):.2f} (offset: {float(time_offset):.2f}s)")
            
            # Create the mask for slicing
            mask = (synchronized_df[actual_timestamp_col] >= adjusted_start) & \
                   (synchronized_df[actual_timestamp_col] <= adjusted_end)
            trial_data = synchronized_df[mask].copy()
            
            if trial_data.empty:
                logger.warning(f"No data found for adjusted time range {float(adjusted_start):.2f} - {float(adjusted_end):.2f}")
                
                # Try to find the closest available data
                closest_start_idx = (synchronized_df[actual_timestamp_col] - adjusted_start).abs().idxmin()
                closest_end_idx = (synchronized_df[actual_timestamp_col] - adjusted_end).abs().idxmin()
                
                logger.info(f"Closest start time: {float(synchronized_df.loc[closest_start_idx, actual_timestamp_col]):.2f}")
                logger.info(f"Closest end time: {float(synchronized_df.loc[closest_end_idx, actual_timestamp_col]):.2f}")
                
                return pd.DataFrame()
            else:
                logger.info(f"Successfully sliced {len(trial_data)} samples for trial")
                
                # Validate that we have reasonable trial duration
                actual_duration = float(trial_data[actual_timestamp_col].max() - trial_data[actual_timestamp_col].min())
                expected_duration = float(routine_end - routine_start)
                
                if abs(actual_duration - expected_duration) > expected_duration * 0.1:  # More than 10% difference
                    logger.warning(f"Trial duration mismatch: expected {expected_duration:.2f}s, got {actual_duration:.2f}s")
                
            return trial_data
            
        except Exception as e:
            logger.error(f"Error slicing trial data: {e}")
            return pd.DataFrame()
    
    def process_trial(self, merged_df: pd.DataFrame, trial_info: pd.Series, 
                     eeg_columns: List[str], gsr_columns: List[str], time_offset: float = 0.0) -> Dict[str, any]:
        """Process a single trial and compute features."""
        
        # Extract trial information with multiple possible column names
        routine_start = None
        routine_end = None
        trial_key = None
        difficulty = None
        
        # Try different possible column names for routine start/end
        for start_col in ['routineStart', 'routine_start', 'start_time', 'StartTime']:
            if start_col in trial_info.index and pd.notna(trial_info[start_col]):
                routine_start = trial_info[start_col]
                break
        
        for end_col in ['routineEnd', 'routine_end', 'end_time', 'EndTime']:
            if end_col in trial_info.index and pd.notna(trial_info[end_col]):
                routine_end = trial_info[end_col]
                break
        
        # Try different possible column names for trial identifier
        for key_col in ['Key', 'key', 'trial', 'Trial', 'trial_id']:
            if key_col in trial_info.index and pd.notna(trial_info[key_col]):
                trial_key = trial_info[key_col]
                break
        
        # Try different possible column names for difficulty/label
        for diff_col in ['Difficulty', 'difficulty', 'label', 'Label', 'condition']:
            if diff_col in trial_info.index and pd.notna(trial_info[diff_col]):
                difficulty = trial_info[diff_col]
                break
        
        # Debug: Show what we found
        logger.debug(f"Trial info columns: {trial_info.index.tolist()}")
        logger.debug(f"Found - start: {routine_start}, end: {routine_end}, key: {trial_key}, difficulty: {difficulty}")
        
        # Convert to numeric if they're strings
        try:
            routine_start = float(routine_start) if routine_start is not None else None
            routine_end = float(routine_end) if routine_end is not None else None
        except (ValueError, TypeError):
            logger.warning(f"Cannot convert routine times to numeric: start={routine_start}, end={routine_end}")
            routine_start = None
            routine_end = None
        
        if routine_start is None or routine_end is None:
            logger.warning(f"Missing or invalid routine start/end times for trial {trial_key}")
            return None
        
        # Slice trial data with proper synchronization
        trial_data = self.slice_trial_data(merged_df, routine_start, routine_end, time_offset)
        
        if trial_data.empty:
            logger.warning(f"No data available for trial {trial_key}")
            return None
        
        # Align EEG and GSR modalities if both are present
        if len(eeg_columns) > 0 or len(gsr_columns) > 0:
            logger.info(f"Aligning modalities for trial {trial_key}")
            trial_data = self.align_modalities(trial_data, eeg_columns, gsr_columns)
        
        # Initialize feature dictionary with synchronization info
        #sampling_rate = self.estimate_sampling_rate(trial_data[trial_data.columns[trial_data.columns.str.contains('timestamp|time', case=False, na=False)].iloc[0]] if any(trial_data.columns.str.contains('timestamp|time', case=False, na=False)) else trial_data.index.to_series())
        # Find timestamp column for sampling rate estimation
        timestamp_col = None
        for col in ['timestamp', 'time', 'Timestamp', 'Time', 'sample_time']:
            if col in trial_data.columns:
                timestamp_col = col
                break

        if timestamp_col:
            sampling_rate = self.estimate_sampling_rate(trial_data[timestamp_col])
        else:
            sampling_rate = self.target_sampling_rate

        features = {
            'trial': trial_key,
            'label': difficulty,
            'routine_start': routine_start,
            'routine_end': routine_end,
            'trial_duration': routine_end - routine_start,
            'n_samples': len(trial_data),
            'sampling_rate': sampling_rate,
            'data_quality_score': self.calculate_data_quality_score(trial_data, eeg_columns, gsr_columns)
        }
        
        # Compute EEG features
        for eeg_col in eeg_columns:
            if eeg_col in trial_data.columns:
                eeg_features = self.compute_basic_features(trial_data[eeg_col], f"EEG_{eeg_col}")
                features.update(eeg_features)
        
        # Compute GSR features
        for gsr_col in gsr_columns:
            if gsr_col in trial_data.columns:
                gsr_features = self.compute_basic_features(trial_data[gsr_col], f"GSR_{gsr_col}")
                features.update(gsr_features)
        
        # Compute overall EEG and GSR statistics
        if eeg_columns:
            available_eeg_cols = [col for col in eeg_columns if col in trial_data.columns]
            if available_eeg_cols:
                eeg_data = trial_data[available_eeg_cols]
                overall_eeg_features = self.compute_basic_features(
                    eeg_data.values.flatten(), "EEG_overall"
                )
                features.update(overall_eeg_features)
                
                # Compute EEG band features if requested
                band_features = self.compute_eeg_band_features(eeg_data)
                features.update(band_features)
        
        if gsr_columns:
            available_gsr_cols = [col for col in gsr_columns if col in trial_data.columns]
            if available_gsr_cols:
                gsr_data = trial_data[available_gsr_cols]
                overall_gsr_features = self.compute_basic_features(
                    gsr_data.values.flatten(), "GSR_overall"
                )
                features.update(overall_gsr_features)
        
        logger.info(f"Successfully processed trial {trial_key} with {len(trial_data)} samples (SR: {sampling_rate:.2f} Hz)")
        return features
    
    def calculate_data_quality_score(self, trial_data: pd.DataFrame, eeg_columns: List[str], 
                                   gsr_columns: List[str]) -> float:
        """Calculate a data quality score for the trial (0-1 scale, 1 = best quality)."""
        
        if trial_data.empty:
            return 0.0
        
        quality_factors = []
        
        # Factor 1: Data completeness (no NaN values)
        total_values = trial_data[eeg_columns + gsr_columns].size
        non_nan_values = trial_data[eeg_columns + gsr_columns].count().sum()
        completeness = non_nan_values / total_values if total_values > 0 else 0.0
        quality_factors.append(completeness)
        
        # Factor 2: Signal stability (low variance in differences indicates stable signal)
        stability_scores = []
        for col in eeg_columns + gsr_columns:
            if col in trial_data.columns and len(trial_data[col].dropna()) > 1:
                signal = trial_data[col].dropna()
                signal_diff = signal.diff().dropna()
                if len(signal_diff) > 0:
                    # Normalize by signal range to get relative stability
                    signal_range = signal.max() - signal.min()
                    if signal_range > 0:
                        stability = 1.0 / (1.0 + signal_diff.std() / signal_range)
                        stability_scores.append(stability)
        
        if stability_scores:
            quality_factors.append(np.mean(stability_scores))
        
        # Factor 3: Temporal consistency (regular sampling)
        timestamp_col = None
        for col in ['timestamp', 'time', 'Timestamp', 'Time']:
            if col in trial_data.columns:
                timestamp_col = col
                break
        
        if timestamp_col and len(trial_data) > 2:
            time_diffs = trial_data[timestamp_col].diff().dropna()
            if len(time_diffs) > 0:
                # Higher score for more consistent time differences
                consistency = 1.0 / (1.0 + time_diffs.std() / time_diffs.mean() if time_diffs.mean() > 0 else 1.0)
                quality_factors.append(consistency)
        
        # Combine all quality factors
        overall_quality = np.mean(quality_factors) if quality_factors else 0.5
        return min(1.0, max(0.0, overall_quality))
    
    def process_participant(self, participant_id: int) -> List[Dict[str, any]]:
        """Process all trials for a single participant."""
        logger.info(f"Processing participant {participant_id}")
        
        # Load files
        merged_df, psy_df = self.load_files(participant_id)
        
        if merged_df is None or psy_df is None:
            logger.warning(f"Skipping participant {participant_id} - missing files")
            return []
        
        # Debug: Show merged data columns and timestamp info
        logger.debug(f"Merged data columns: {list(merged_df.columns)}")
        
        # Calculate time alignment offset
        timestamp_col = None
        for col in ['timestamp', 'time', 'Timestamp', 'Time', 'sample_time']:
            if col in merged_df.columns:
                timestamp_col = col
                break
        
        time_offset = 0.0
        if timestamp_col:
            time_offset = self.find_time_alignment(merged_df, psy_df, timestamp_col)
        
        # Identify EEG and GSR columns
        eeg_columns, gsr_columns = self.identify_eeg_gsr_columns(merged_df)
        
        if not eeg_columns and not gsr_columns:
            logger.warning(f"No EEG or GSR columns found for participant {participant_id}")
            return []
        
        # Process each trial
        trial_features = []
        
        for idx, trial_info in psy_df.iterrows():
            logger.debug(f"Processing trial {idx}")
            features = self.process_trial(merged_df, trial_info, eeg_columns, gsr_columns, time_offset)
            
            if features is not None:
                features['participant'] = participant_id
                trial_features.append(features)
            else:
                logger.warning(f"Failed to process trial {idx} for participant {participant_id}")
        
        logger.info(f"Successfully processed {len(trial_features)} trials for participant {participant_id}")
        return trial_features
    
    def run_feature_extraction(self, participant_range: range = range(1, 39)) -> pd.DataFrame:
        """Run feature extraction for all participants."""
        logger.info("Starting feature extraction process")
        
        all_trial_features = []
        
        for participant_id in participant_range:
            try:
                participant_features = self.process_participant(participant_id)
                all_trial_features.extend(participant_features)
            except Exception as e:
                logger.error(f"Error processing participant {participant_id}: {e}")
                continue
        
        if not all_trial_features:
            logger.error("No features extracted for any participant")
            return pd.DataFrame()
        
        # Convert to DataFrame
        feature_df = pd.DataFrame(all_trial_features)
        
        # Reorder columns for better readability
        priority_columns = ['participant', 'trial', 'label', 'routine_start', 'routine_end', 
                          'trial_duration', 'n_samples']
        other_columns = [col for col in feature_df.columns if col not in priority_columns]
        ordered_columns = priority_columns + sorted(other_columns)
        
        feature_df = feature_df[[col for col in ordered_columns if col in feature_df.columns]]
        
        logger.info(f"Feature extraction complete. Final dataset shape: {feature_df.shape}")
        logger.info(f"Features extracted: {len(feature_df.columns)} columns")
        
        return feature_df
    
    def save_dataset(self, feature_df: pd.DataFrame, filename: str = "per_trial_dataset.csv"):
        """Save the per-trial dataset."""
        output_path = self.output_dir / filename
        feature_df.to_csv(output_path, index=False)
        logger.info(f"Dataset saved to: {output_path}")
        
        # Print summary statistics
        print("\n" + "="*60)
        print("FEATURE EXTRACTION SUMMARY")
        print("="*60)
        print(f"Total trials processed: {len(feature_df)}")
        print(f"Number of participants: {feature_df['participant'].nunique()}")
        print(f"Number of features: {len(feature_df.columns)}")
        print(f"Labels distribution:")
        if 'label' in feature_df.columns:
            print(feature_df['label'].value_counts())
        print(f"Output saved to: {output_path}")


def main():
    """Main function to run the feature extraction with timestamp synchronization."""
    
    # Initialize the feature extractor with target sampling rate
    extractor = EEGGSRFeatureExtractor(
        merged_data_dir="../data/merged_data",
        psy_data_dir="../data/PSY",
        output_dir="../data",
        target_sampling_rate=250.0  # Set target sampling rate for alignment
    )
    
    # Run feature extraction for participants 1-38
    feature_df = extractor.run_feature_extraction(range(1, 39))
    
    if not feature_df.empty:
        # Save the dataset
        extractor.save_dataset(feature_df)
        
        # Display first few rows
        print("\nFirst 5 rows of the dataset:")
        print(feature_df.head())
        
        print("\nDataset info:")
        print(feature_df.info())
    else:
        logger.error("No data was processed successfully")


if __name__ == "__main__":
    main()
```
